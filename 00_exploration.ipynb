```python
# Notebook 00 - Exploration
# Load the 3 CSV files from Bronze layer

BRONZE_PATH = "/Volumes/insurance_project/bronze/raw_data"

df_insurance = spark.read.csv(f"{BRONZE_PATH}/insurance_data.csv", header=True, inferSchema=True)
df_employees = spark.read.csv(f"{BRONZE_PATH}/employee_data.csv", header=True, inferSchema=True)
df_vendors = spark.read.csv(f"{BRONZE_PATH}/vendor_data.csv", header=True, inferSchema=True)

print("Files loaded successfully!")
```

```python
# Check the first rows of each dataset

print("=== INSURANCE DATA ===")
df_insurance.show(3)

print("=== EMPLOYEE DATA ===")
df_employees.show(3)

print("=== VENDOR DATA ===")
df_vendors.show(3)
```

```python
# Check the schema of each dataset

print("=== INSURANCE DATA ===")
df_insurance.printSchema()

print("=== EMPLOYEE DATA ===")
df_employees.printSchema()

print("=== VENDOR DATA ===")
df_vendors.printSchema()
```

```python
# Count rows and columns for each dataset

print("insurance_data:", df_insurance.count(), "rows |", len(df_insurance.columns), "columns")
print("employee_data: ", df_employees.count(), "rows |", len(df_employees.columns), "columns")
print("vendor_data: ", df_vendors.count(), "rows |", len(df_vendors.columns), "columns")
```

```python
# Check nulls in key columns

from pyspark.sql.functions import col, sum as spark_sum

key_columns = ["CUSTOMER_ID", "POLICY_NUMBER", "INSURANCE_TYPE", 
               "CLAIM_AMOUNT", "AGENT_ID", "VENDOR_ID"]

df_insurance.select([
    spark_sum(col(c).isNull().cast("int")).alias(c) for c in key_columns
]).show()
```
