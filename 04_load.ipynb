```python
# Notebook 04 - Load
# Save Gold tables as Delta Tables in Unity Catalog

GOLD_PATH = "/Volumes/insurance_project/gold/star_schema"

dim_customers = spark.read.parquet(f"{GOLD_PATH}/dim_customers")
dim_policies = spark.read.parquet(f"{GOLD_PATH}/dim_policies")
dim_agents = spark.read.parquet(f"{GOLD_PATH}/dim_agents")
dim_vendors = spark.read.parquet(f"{GOLD_PATH}/dim_vendors")
dim_dates = spark.read.parquet(f"{GOLD_PATH}/dim_dates")
fact_claims = spark.read.parquet(f"{GOLD_PATH}/fact_claims")

print("Gold data loaded!")
```

```python
# Save all tables as Delta Tables in Unity Catalog

dim_customers.write.format("delta").mode("overwrite").saveAsTable("insurance_project.gold.dim_customers")
dim_policies.write.format("delta").mode("overwrite").saveAsTable("insurance_project.gold.dim_policies")
dim_agents.write.format("delta").mode("overwrite").saveAsTable("insurance_project.gold.dim_agents")
dim_vendors.write.format("delta").mode("overwrite").saveAsTable("insurance_project.gold.dim_vendors")
dim_dates.write.format("delta").mode("overwrite").saveAsTable("insurance_project.gold.dim_dates")
fact_claims.write.format("delta").mode("overwrite").saveAsTable("insurance_project.gold.fact_claims")

print("Delta Tables saved successfully!")
```
