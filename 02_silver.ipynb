```python
# Notebook 02 - Silver
# Load data from Bronze layer and clean it

BRONZE_PATH = "/Volumes/insurance_project/bronze/raw_data"

df_insurance = spark.read.csv(f"{BRONZE_PATH}/insurance_data.csv", header=True, inferSchema=True)
df_employees = spark.read.csv(f"{BRONZE_PATH}/employee_data.csv", header=True, inferSchema=True)
df_vendors = spark.read.csv(f"{BRONZE_PATH}/vendor_data.csv", header=True, inferSchema=True)

print("Bronze data loaded!")
```

```python
# Clean all datasets

from pyspark.sql.functions import col

# insurance_data - fix data types, remove duplicates and nulls in key columns

df_insurance_clean = df_insurance.dropDuplicates()
df_insurance_clean = df_insurance_clean.dropna(subset=["CUSTOMER_ID", "POLICY_NUMBER", "INSURANCE_TYPE"])
df_insurance_clean = df_insurance_clean.withColumn("CLAIM_AMOUNT", col("CLAIM_AMOUNT").cast("double"))
df_insurance_clean = df_insurance_clean.withColumn("ROUTING_NUMBER", col("ROUTING_NUMBER").cast("string"))

# employee_data - fix data types, remove duplicates and nulls in key columns

df_employees_clean = df_employees.dropDuplicates()
df_employees_clean = df_employees_clean.dropna(subset=["AGENT_ID"])
df_employees_clean = df_employees_clean.withColumn("EMP_ROUTING_NUMBER", col("EMP_ROUTING_NUMBER").cast("string"))

# vendor_data - remove duplicates and nulls in key columns

df_vendors_clean = df_vendors.dropDuplicates()
df_vendors_clean = df_vendors_clean.dropna(subset=["VENDOR_ID"])

print("Data cleaned successfully!")
```

```python
# Compare row counts before and after cleaning

print("=== ROW COUNTS: Before vs After cleaning ===")
print("insurance_data:", df_insurance.count(), "→", df_insurance_clean.count())
print("employee_data:", df_employees.count(), "→", df_employees_clean.count())
print("vendor_data:", df_vendors.count(), "→", df_vendors_clean.count())
```

```python
# Save cleaned data to Silver layer

SILVER_PATH = "/Volumes/insurance_project/silver/clean_data"

df_insurance_clean.write.mode("overwrite").parquet(f"{SILVER_PATH}/insurance_data")
df_employees_clean.write.mode("overwrite").parquet(f"{SILVER_PATH}/employee_data")
df_vendors_clean.write.mode("overwrite").parquet(f"{SILVER_PATH}/vendor_data")

print("Silver layer saved successfully!")
```
